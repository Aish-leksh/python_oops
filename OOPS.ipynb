Initialize a class person with constructor to initialize the name and the aadhaar number, children and age should set as 0. you have to define get_name, to print the name field, define get_adhaar() to print aadhaar number, set_age to assign a new value to the age field also define senior_citizen() function which checks for the age>60.Finally create an object for this class and call the respective methods.

```python
class person:
    def __init__(self,name,aadhaar):
        self.name=name
        self.aadhaar=aadhaar
        self.children=0
        self.age=0
    def get_name(self):
        print(self.name)
    def get_adhaar(self):
        print(self.aadhaar)
    def set_age(self,age):
        self.age=age
    def senior_citizen(self):
      if self.age>60:
        print("Senior citizen")
      else:
        print("Not senior citizen")
name =input("Enter your name:")
aadhaar=int(input("Enter the aadhaar number:"))
age=int(input("Enter your age:"))
p1=person(name,aadhaar)
p1.set_age(age)
p1.senior_citizen()
p1.get_name()
p1.get_adhaar()
```

Create a class circle with the constructor initializing the radius r, another function area to calculate the area of the circle, another function circumference to calculate the circumference of the circle.

```python
class circle:
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14*self.r*self.r
    def circumference(self):
        return 2*3.14*self.r
r=int(input("Enter the radius:"))
c1=circle(r)
print("Area of the circle:",c1.area())
print("Circumference of the circle:",c1.circumference())
```

There are two classes namely num1 and num2, In num1 class there are two variables a and b that needs to be initialized using constructor it also have a member function, show_values() to print this value. num2 class is inherited from num1 and it has two functions. first one is show_sum() that display the sum of the variables the second one is show_product() that display the product of the variables.

```python
class num1:
  def __init__(self,a,b):
    self.a=a
    self.b=b
  def show_values(self):
    print("a:",self.a)
    print("b:",self.b)
class num2(num1):
  def show_sum(self):
    print("Sum:",self.a+self.b)
  def show_product(self):
    print("Product:",self.a*self.b)
a=int(input("Enter the value of num1:"))
b=int(input("Enter the value of num2:"))
n=num2(a,b)
n.show_values
n.show_sum()
n.show_product()
```

There are two classes student and master_student. Student is the base class and master_student is the derived class. the class student has two member variables ID and mark. There is a constructor to initialize these variables and a function show result which will check the mark for passed and failed. If the mark is >= 40, It is a pass otherwise it is a fail. similarly in the class master_student contain a function show result that will check wheather the mark is >= 50. If that is the case then it is a pass otherwise it will be a fail . Initialize the student object and a master student object and call the functions respectively.  

```python
class Student:
   def __init__(self,i,m):
     self.ID=i
     self.mark=m
   def show_result(self):
     if self.mark>=40:
      print("Pass")
     else:
      print("Fail")
class master_student(Student):
  def show_result(self):
    if self.mark>=50:
      print("Pass")
    else:
      print("Fail")
m=int(input("Enter the mark:"))
i=int(input("Enter the ID:"))
s1=Student(i,m)
s1.show_result()
m=int(input("Enter the mark:"))
i=int(input("Enter the ID:"))
s2=master_student(i,m)
s2.show_result()

```
